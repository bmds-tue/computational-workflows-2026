#### What did change?

The second run was much faster (1m 54s instead of 7m 44s). All processes were executed again, but Nextflow did not need to download the Docker images anymore, because they were already stored locally after the first run.

```python
# now set -resume to the command. What did change?

!nextflow run nf-core/differentialabundance -r 2.0.0 -profile test,docker --outdir results -resume 
```

#### What did change using -resume?

With -resume it continoues the last run and uses some computation from the previous run. 
The runtime is 16.1s. So it did not start from scratch and therefore is much faster

## Lets look at the results

### What is differential abundance analysis?

Differential abundance analysis is a statistical method to identify features (e.g genes , transcripts, proteins) whose abundance differs significantly between conditions such as treated vs untreaded or healthy vs control

##### Give the most important plots from the report:

![Volcano-Plot](images/volcano_treatment.png)

The volcano plot compares hND6 and mCherry. Each point is a gene. The x-axis shows how much the expression changes (log2 fold change), and the y-axis shows how significant the change is. Most genes are close to 0 and below the significance line, so they do not change. Only a few genes are significant (differential status = true) and all of them are upregulated. No genes are significantly downregulated. 

![PCA-Plot](images/pca_treatment.png)


The PCA plot shows six samples -> 3 mCherry and 3 hND6

PC1 explains 45.3% of the variance, but it does not separate the two groups. Instead, the samples form pairs along PC1 (381/385, 382/386, 383/384). This suggests that the biggest differences come from other factors, not from the treatment itself. 

PC2 explains 22.1% of the variance. The two groups separate along PC2 -> all mCherry samples are at the top and all hND6 samples are at the bottom. So the treatment has an effect, but but explains less than possible batch effect.
